# Python对象详解

Python是一种面向对象的编程语言，理解Python对象是掌握Python编程的关键。本教程将详细介绍Python对象的相关知识，包括对象的创建、属性、方法、继承等核心概念。

## 1. Python对象基础概念

在Python中，一切皆为对象。每个对象都有以下特征：
- **身份（Identity）**：对象的唯一标识符，类似内存地址
- **类型（Type）**：对象的数据类型，决定了对象可以执行的操作
- **值（Value）**：对象所包含的数据

```python
# 让我们先了解对象的基本属性
num = 42
text = "Hello, Python!"
lst = [1, 2, 3]

print("=== 数字对象 ===")
print(f"值: {num}")
print(f"类型: {type(num)}")
print(f"身份(id): {id(num)}")
print(f"是否为int类型: {isinstance(num, int)}")

print("\n=== 字符串对象 ===")
print(f"值: {text}")
print(f"类型: {type(text)}")
print(f"身份(id): {id(text)}")
print(f"长度: {len(text)}")

print("\n=== 列表对象 ===")
print(f"值: {lst}")
print(f"类型: {type(lst)}")
print(f"身份(id): {id(lst)}")
print(f"长度: {len(lst)}")
```

## 3. 类的定义和对象创建

```python
# 定义一个简单的类
class Person:
    """人类：包含姓名和年龄属性"""
    
    # 类变量（所有实例共享）
    species = "Homo sapiens"
    
    def __init__(self, name, age):
        """初始化方法，创建对象时自动调用"""
        self.name = name  # 实例变量
        self.age = age
    
    def introduce(self):
        """实例方法"""
        return f"我是{self.name}，今年{self.age}岁"
    
    def have_birthday(self):
        """增加年龄"""
        self.age += 1
        return f"{self.name}生日快乐！现在是{self.age}岁"
    
    @classmethod #类方法，第一个参数是类本身，而不是实例
    def get_species(cls):
        """类方法"""
        return f"人类属于{cls.species}物种"
    
    @staticmethod #静态方法，没有self参数，不能访问类变量和实例变量
    def is_adult(age):
        """静态方法"""
        return age >= 18

# 创建对象
person1 = Person("张三", 25)
person2 = Person("李四", 17)

print("=== 对象操作 ===")
print(person1.introduce())
print(person2.introduce())
print(person1.have_birthday())
print(f"\n{person1.name}是成年人吗？{Person.is_adult(person1.age)}")
print(f"{person2.name}是成年人吗？{Person.is_adult(person2.age)}")
print(f"\n{Person.get_species()}")
```

## 4. 对象的属性和方法详解

```python
# 属性的增删改查
class Student:
    def __init__(self, name):
        self.name = name
        self._private_attr = "私有属性"  # 约定以单下划线开头
        self.__very_private = "强私有属性"  # 约定以双下划线开头
    
    def show_info(self):
        return f"姓名: {self.name}, 私有属性: {self._private_attr}"

student = Student("王五")

print("=== 访问属性 ===")
print(student.name)  # 访问公有属性
print(student._private_attr)  # 访问私有属性（约定不直接访问）
# print(student.__very_private)  # 这会报错
print(student._Student__very_private)  # 访问强私有属性的方式（不推荐）

print("\n=== 修改属性 ===")
student.name = "王六"
print(f"修改后的姓名: {student.name}")

print("\n=== 动态添加属性 ===")
student.grade = "三年级"
student.score = 95
print(f"年级: {student.grade}")
print(f"分数: {student.score}")

print("\n=== 查看对象的所有属性 ===")
print(f"所有属性: {vars(student)}")
print(f"所有属性（包括方法）: {dir(student)}")

print("\n=== 删除属性 ===")
del student.score
print(f"删除score后: {vars(student)}")
```

## 5. 特殊方法（魔术方法）

```python
# 自定义类的特殊方法
class Vector:
    """二维向量类"""
    
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __str__(self):
        """返回对象的字符串表示"""
        return f"Vector({self.x}, {self.y})"
    
    def __repr__(self):
        """返回对象的官方字符串表示"""
        return f"Vector(x={self.x}, y={self.y})"
    
    def __add__(self, other):
        """向量加法"""
        if isinstance(other, Vector): # 判断other是否是Vector类型的对象
            return Vector(self.x + other.x, self.y + other.y)
        return NotImplemented
    
    def __sub__(self, other):
        """向量减法"""
        if isinstance(other, Vector):
            return Vector(self.x - other.x, self.y - other.y)
        return NotImplemented
    
    def __mul__(self, scalar):
        """向量与标量相乘"""
        if isinstance(scalar, (int, float)):
            return Vector(self.x * scalar, self.y * scalar)
        return NotImplemented
    
    def __eq__(self, other):
        """判断两个向量是否相等"""
        if isinstance(other, Vector):
            return self.x == other.x and self.y == other.y
        return False
    
    def __len__(self):
        """返回向量的模长"""
        return int((self.x**2 + self.y**2)**0.5)
    
    def __getitem__(self, index):
        """支持下标访问"""
        if index == 0:
            return self.x
        elif index == 1:
            return self.y
        else:
            raise IndexError("Vector index out of range")

# 测试特殊方法
v1 = Vector(3, 4)
v2 = Vector(1, 2)

print("=== 特殊方法测试 ===")
print(f"v1 = {v1}")  # 调用__str__
print(f"v1的官方表示: {repr(v1)}")  # 调用__repr__

v3 = v1 + v2  # 调用__add__
print(f"v1 + v2 = {v3}")

v4 = v1 - v2  # 调用__sub__
print(f"v1 - v2 = {v4}")

v5 = v1 * 2  # 调用__mul__
print(f"v1 * 2 = {v5}")

print(f"v1 == v2: {v1 == v2}")  # 调用__eq__
print(f"v1的长度: {len(v1)}")  # 调用__len__
print(f"v1[0] = {v1[0]}, v1[1] = {v1[1]}")  # 调用__getitem__
```

## 6. 继承和多态

```python
# 基类（父类）
class Animal:
    """动物基类"""
    
    def __init__(self, name, species):
        self.name = name
        self.species = species
    
    def speak(self):
        """动物发出声音"""
        raise NotImplementedError("子类必须实现speak方法")
    
    def info(self):
        """返回动物信息"""
        return f"{self.name}是一只{self.species}"

# 派生类（子类）
class Dog(Animal):
    """狗类"""
    
    def __init__(self, name, breed):
        super().__init__(name, "狗")  # 调用父类构造函数
        self.breed = breed  # 新增属性
    
    def speak(self):
        """重写父类方法"""
        return "汪汪！"
    
    def info(self):
        """重写父类方法"""
        return f"{self.name}是一只{self.breed}品种的{self.species}"
    
    def wag_tail(self):
        """新增方法"""
        return f"{self.name}摇着尾巴"

class Cat(Animal):
    """猫类"""
    
    def __init__(self, name, color):
        super().__init__(name, "猫")
        self.color = color
    
    def speak(self):
        return "喵喵！"
    
    def info(self):
        return f"{self.name}是一只{self.color}色的{self.species}"

# 多重继承
class Pet(Dog):
    """宠物类，继承自Dog"""
    
    def __init__(self, name, breed, owner):
        super().__init__(name, breed)
        self.owner = owner
    
    def play(self):
        return f"{self.name}和主人{self.owner}一起玩耍"

# 测试继承和多态
print("=== 继承和多态 ===")
dog = Dog("小白", "金毛")
cat = Cat("咪咪", "橘色")
pet_dog = Pet("小黑", "拉布拉多", "小明")

print(dog.info())
print(f"{dog.name}说: {dog.speak()}")
print(f"{dog.wag_tail()}")

print(f"\n{cat.info()}")
print(f"{cat.name}说: {cat.speak()}")

print(f"\n{pet_dog.info()}")
print(f"{pet_dog.play()}")

print("\n=== 多态测试 ===")
animals = [dog, cat]
for animal in animals:
    print(f"{animal.name}: {animal.speak()}")  # 调用各自的方法

print(f"\n=== 方法解析顺序(MRO) ===")
print(f"Pet类的MRO: {[cls.__name__ for cls in Pet.__mro__]}")
```

## 7. 对象的比较和哈希

```python
# 比较和哈希的实现
class Book:
    """图书类"""
    
    def __init__(self, title, author, isbn):
        self.title = title
        self.author = author
        self.isbn = isbn
    
    def __eq__(self, other):
        """相等比较"""
        if not isinstance(other, Book):
            return False
        return self.isbn == other.isbn  # 以ISBN作为唯一标识
    
    def __hash__(self):
        """哈希值计算"""
        return hash(self.isbn)
    
    def __str__(self):
        return f"《{self.title}》- {self.author}"

# 测试比较和哈希
book1 = Book("Python编程", "张三", "978-7-111-11111-1")
book2 = Book("Python编程", "李四", "978-7-111-11111-1")  # 相同ISBN
book3 = Book("Java编程", "王五", "978-7-111-11111-2")

print("=== 对象比较 ===")
print(f"book1 == book2: {book1 == book2}")
print(f"book1 == book3: {book1 == book3}")
print(f"book1 == '字符串': {book1 == '字符串'}")  # 不同类型

print("\n=== 对象哈希 ===")
print(f"book1的哈希值: {hash(book1)}")
print(f"book2的哈希值: {hash(book2)}")
print(f"book3的哈希值: {hash(book3)}")

# 可哈希的对象可以作为字典的键或集合的元素
print("\n=== 哈希的实际应用 ===")
book_set = {book1, book2, book3}  # 集合去重
print(f"图书集合大小: {len(book_set)}")  # 因为book1和book2有相同ISBN，所以只保留一个

book_dict = {
    book1: "图书馆A",
    book2: "图书馆B",  # 这个会覆盖book1的条目，因为ISBN相同
    book3: "图书馆C"
}
print(f"图书字典大小: {len(book_dict)}")
for book, location in book_dict.items():
    print(f"{book}: {location}")
```

## 8. 对象的复制

```python
import copy

class Company:
    """公司类"""
    
    def __init__(self, name, employees):
        self.name = name
        self.employees = employees  # 列表类型
    
    def __str__(self):
        return f"{self.name}: {self.employees}"

# 测试不同复制方式
company1 = Company("科技公司", ["张三", "李四", "王五"])

print("=== 浅拷贝 vs 深拷贝 ===")
print(f"原对象: {company1}")

# 浅拷贝
shallow_copy = copy.copy(company1)
deep_copy = copy.deepcopy(company1)

print(f"浅拷贝: {shallow_copy}")
print(f"深拷贝: {deep_copy}")

# 修改原对象的嵌套对象
company1.employees.append("赵六")

print(f"\n修改原对象后:")
print(f"原对象: {company1}")
print(f"浅拷贝: {shallow_copy}")  # 会受影响
print(f"深拷贝: {deep_copy}")  # 不受影响

# 自定义拷贝方法
class Person:
    """人员类"""
    
    def __init__(self, name, address):
        self.name = name
        self.address = address  # 嵌套对象
    
    def __copy__(self):
        """自定义浅拷贝"""
        new_person = Person(self.name, self.address)
        return new_person
    
    def __deepcopy__(self, memo):
        """自定义深拷贝"""
        new_person = Person(self.name, copy.deepcopy(self.address, memo))
        return new_person
    
    def __str__(self):
        return f"{self.name} - {self.address}"

class Address:
    """地址类"""
    
    def __init__(self, city, street):
        self.city = city
        self.street = street
    
    def __str__(self):
        return f"{self.city} {self.street}"

# 测试自定义拷贝
addr = Address("北京", "中关村大街1号")
person1 = Person("张三", addr)

print(f"\n=== 自定义拷贝测试 ===")
print(f"原对象: {person1}")

person2 = copy.deepcopy(person1)
print(f"深拷贝对象: {person2}")

# 修改原对象的嵌套对象
person1.address.city = "上海"
print(f"\n修改原对象后:")
print(f"原对象: {person1}")
print(f"深拷贝对象: {person2}")  # 不受影响
```

## 9. 属性描述符

```python
# 属性描述符示例
class Positive:
    """正数描述符"""
    
    def __init__(self, name):
        self.name = name
        self._attr_name = f"_{name}"
    
    def __get__(self, obj, objtype=None):
        """获取属性值"""
        return getattr(obj, self._attr_name, 0)
    
    def __set__(self, obj, value):
        """设置属性值"""
        if value <= 0:
            raise ValueError(f"{self.name}必须为正数")
        setattr(obj, self._attr_name, value)
    
    def __delete__(self, obj):
        """删除属性"""
        delattr(obj, self._attr_name)

class Product:
    """商品类"""
    
    # 使用描述符
    price = Positive('price')
    quantity = Positive('quantity')
    
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity
    
    @property # 使用property装饰器定义只读属性
    def total_price(self):
        """计算总价"""
        return self.price * self.quantity
    
    def __str__(self):
        return f"{self.name}: 单价{self.price}, 数量{self.quantity}, 总价{self.total_price}"

# 测试描述符
print("=== 属性描述符测试 ===")
product = Product("笔记本电脑", 5999, 5)
print(product)

print(f"\n尝试设置负数价格:")
try:
    product.price = -100
except ValueError as e:
    print(f"错误: {e}")

print(f"\n尝试设置负数数量:")
try:
    product.quantity = 0
except ValueError as e:
    print(f"错误: {e}")

print(f"\n修改价格和数量:")
product.price = 4999
product.quantity = 10
print(product)
```

## 10. 实际应用示例

```python
# 实际应用：简单的银行账户系统
class BankAccount:
    """银行账户类"""
    
    # 类变量
    bank_name = "中国银行"
    total_accounts = 0
    
    def __init__(self, owner_name, initial_balance=0):
        self.owner_name = owner_name
        self._balance = initial_balance  # 私有属性
        self._transaction_history = []  # 交易历史
        BankAccount.total_accounts += 1
        self.account_id = BankAccount.total_accounts
        self._record_transaction("开户", initial_balance)
    
    def deposit(self, amount):
        """存款"""
        if amount <= 0:
            raise ValueError("存款金额必须为正数")
        self._balance += amount
        self._record_transaction("存款", amount)
        return f"存款成功，当前余额: {self._balance:.2f}"
    
    def withdraw(self, amount):
        """取款"""
        if amount <= 0:
            raise ValueError("取款金额必须为正数")
        if amount > self._balance:
            raise ValueError("余额不足")
        self._balance -= amount
        self._record_transaction("取款", -amount)
        return f"取款成功，当前余额: {self._balance:.2f}"
    
    def transfer(self, other_account, amount):
        """转账"""
        if not isinstance(other_account, BankAccount):
            raise TypeError("转账对象必须是银行账户")
        if amount <= 0:
            raise ValueError("转账金额必须为正数")
        if amount > self._balance:
            raise ValueError("余额不足")
        
        self._balance -= amount
        other_account._balance += amount
        self._record_transaction(f"转账给{other_account.owner_name}", -amount)
        other_account._record_transaction(f"来自{self.owner_name}的转账", amount)
        return f"转账成功"
    
    def _record_transaction(self, description, amount):
        """记录交易历史（私有方法）"""
        import datetime
        timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        self._transaction_history.append({
            "时间": timestamp,
            "描述": description,
            "金额": amount,
            "余额": self._balance
        })
    
    @property
    def balance(self):
        """余额属性（只读）"""
        return self._balance
    
    @property
    def account_info(self):
        """账户信息"""
        return {
            "账户ID": self.account_id,
            "户名": self.owner_name,
            "余额": self._balance,
            "开户银行": self.bank_name,
            "交易次数": len(self._transaction_history)
        }
    
    def show_transaction_history(self, limit=10):
        """显示交易历史"""
        print(f"\n{self.owner_name}的交易历史 (最近{limit}笔):")
        for transaction in self._transaction_history[-limit:]:
            print(f"{transaction['时间']} - {transaction['描述']}: {transaction['金额']:+.2f} (余额: {transaction['余额']:.2f})")
    
    def __str__(self):
        return f"账户{self.account_id}: {self.owner_name}, 余额: {self._balance:.2f}"
    
    def __repr__(self):
        return f"BankAccount(owner='{self.owner_name}', balance={self._balance})"

# 测试银行账户系统
print("=== 银行账户系统测试 ===")

# 创建账户
account1 = BankAccount("张三", 10000)
account2 = BankAccount("李四", 5000)

print(f"总账户数: {BankAccount.total_accounts}")
print(f"账户1: {account1}")
print(f"账户2: {account2}")

# 存款和取款
print(f"\n{account1.deposit(2000)}")
print(f"{account1.withdraw(1500)}")

# 转账
print(f"\n{account1.transfer(account2, 3000)}")

# 显示交易历史
account1.show_transaction_history()
account2.show_transaction_history()

# 显示账户信息
print(f"\n账户1信息: {account1.account_info}")
print(f"\n最终状态:")
print(f"账户1: {account1}")
print(f"账户2: {account2}")
```
